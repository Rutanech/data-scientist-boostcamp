## 1. Import libraries

## 2. Load data

อ่านไฟล์ CSV เข้ามาเป็นตารางข้อมูลชื่อ `sales`

```python
sales = pd.read_csv('../data/simple_sales.csv')
sales.head()
```

เริ่มสำรวจข้อมูลด้วยคำสั่งพื้นฐาน

## 4. Python basics quick practice

ลองใช้ตัวแปรและ loop แบบง่าย ๆ จากข้อมูลจริง

```python
total_orders = len(sales)
regions = sales['region'].unique()

print('Total orders:', total_orders)
print('Regions:', regions)
```

```python
for region in regions:
    print(region)
```

## 5. Select, filter, and sort

ฝึกเลือก column, กรอง row และเรียงข้อมูล

```python
sales[['order_id', 'region', 'category', 'quantity', 'unit_price']].head()
```

```python
bangkok_sales = sales[sales['region'] == 'Bangkok']
bangkok_sales
```

```python
sales.sort_values('unit_price', ascending=False).head(5)
```

## 6. Create calculated columns

สร้าง column ใหม่เพื่อคำนวณยอดขายก่อนและหลังหักส่วนลด

```python
sales['gross_sales'] = sales['quantity'] * sales['unit_price']
sales['discount_amount'] = sales['gross_sales'] * sales['discount']
sales['net_sales'] = sales['gross_sales'] - sales['discount_amount']

sales.head()
```

## 7. Groupby practice

ใช้ `groupby` เพื่อสรุปยอดขายตามกลุ่มต่าง ๆ

```python
sales.groupby('region')['net_sales'].sum().sort_values(ascending=False)
```

```python
sales.groupby('category')['net_sales'].agg(['count', 'sum', 'mean']).sort_values('sum', ascending=False)
```

```python
sales.groupby(['region', 'category'])['net_sales'].sum().reset_index().sort_values('net_sales', ascending=False)
```

## 8. Mini project questions

ตอบคำถามเหล่านี้ด้วย pandas:

1. Region ไหนมียอดขายสุทธิสูงที่สุด?
2. Category ไหนทำยอดขายสุทธิสูงที่สุด?
3. Product ไหนขายได้จำนวนชิ้นรวมสูงที่สุด?
4. ลูกค้า New หรือ Returning สร้างยอดขายสุทธิมากกว่า?
5. Order ไหนมีส่วนลดมากที่สุด?

```python
# 1. Region ไหนมียอดขายสุทธิสูงที่สุด?
sales.groupby('region')['net_sales'].sum().sort_values(ascending=False)
```

```python
# 2. Category ไหนทำยอดขายสุทธิสูงที่สุด?
sales.groupby('category')['net_sales'].sum().sort_values(ascending=False)
```

```python
# 3. Product ไหนขายได้จำนวนชิ้นรวมสูงที่สุด?
sales.groupby('product')['quantity'].sum().sort_values(ascending=False)
```

```python
# 4. ลูกค้า New หรือ Returning สร้างยอดขายสุทธิมากกว่า?
sales.groupby('customer_type')['net_sales'].sum().sort_values(ascending=False)
```

```python
# 5. Order ไหนมีส่วนลดมากที่สุด?
sales.sort_values('discount_amount', ascending=False).head(5)
```
